```python
from ultralytics import YOLO
import cv2

model = YOLO("/content/drive/MyDrive/SIh dataset aaradhy/Pothole_Model/weights/best.pt")

cap = cv2.VideoCapture("/content/drive/MyDrive/SIh dataset aaradhy/test videos/road test.mp4")

frame_no = 0

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    results = model(frame, conf=0.35)

    count = len(results[0].boxes)

    print(f"Frame {frame_no}: {count} potholes detected")

    frame_no += 1

cap.release()
```
